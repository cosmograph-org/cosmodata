```python
# final prep

def nodes_from_links(links):
    nodes = sorted(set([x['source'] for x in links] + [x['target'] for x in links]))
    return [{'id': x} for x in nodes]

class links_from:
    @staticmethod
    def node_pairs(node_pairs):
        links = [
            {"source": from_, "target": to_} 
            for from_, to_ in node_pairs #zip(*dd['net_upu'].nonzero())
        ]
        return links

    def source_and_target_arrays(source_ids, target_ids):
        return links_from.node_pairs(zip(source_ids, target_ids))
```

```python
import re

def camel_to_snake_case(name):
    # Replace any non-word character with an underscore
    name = re.sub(r'\W+', '_', name)
    # Insert an underscore before any capital letter (except the first one)
    name = re.sub(r'([a-z])([A-Z])', r'\1_\2', name).lower()
    return name

def snake_to_camel_case(name):
    # Split the string into words separated by underscores
    words = name.split('_')
    # Convert each word to title case (i.e., with the first letter capitalized)
    words = [w.capitalize() for w in words]
    # Join the words together into a single string
    return ''.join(words)
```

```python
import io
from functools import partial
import scipy.io

import graze
from dol import Pipe, FilesOfZip
from operator import methodcaller

def next_asserting_uniqueness(iterator):
    v = next(iterator)
    assert next(iterator, None) is None, "There was more than one value in iterator"
    return v

def postget_factory(val_trans_for_name, k, v):
    if k in val_trans_for_name:
        value_transformer = val_trans_for_name[k]
        return value_transformer(v)
    return v

def link_to_data_store(link_store, val_trans_for_name=()):
    val_trans_for_name = dict(val_trans_for_name)
    link_to_data_store_trans = Pipe(
        dol.wrap_kvs(postget=partial(postget_factory, val_trans_for_name)),
        dol.add_ipython_key_completions
    )
    return link_to_data_store_trans(link_store)


_graze = partial(graze.graze, key_ingress=graze.key_ingress_print_downloading_message)
```
